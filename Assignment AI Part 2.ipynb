```python
# Part II: Map Graph Representation
# Dictionary containing all 16 cities, adjacent nodes, and edge weights (distances)

graph = {
    'Seattle': {'San Francisco': 678, 'Chicago': 1737},
    'San Francisco': {'Seattle': 678, 'Los Angeles': 348, 'Riverside': 386},
    'Los Angeles': {'San Francisco': 348, 'Riverside': 50, 'Phoenix': 357},
    'Riverside': {'San Francisco': 386, 'Los Angeles': 50, 'Phoenix': 307, 'Chicago': 1704},
    'Phoenix': {'Los Angeles': 357, 'Riverside': 307, 'Dallas': 887, 'Houston': 1015},
    'Dallas': {'Phoenix': 887, 'Chicago': 805, 'Atlanta': 721, 'Houston': 225},
    'Houston': {'Phoenix': 1015, 'Dallas': 225, 'Atlanta': 702, 'Miami': 968},
    'Chicago': {'Seattle': 1737, 'Riverside': 1704, 'Dallas': 805, 'Detroit': 238, 'Atlanta': 588},
    'Detroit': {'Chicago': 238, 'Boston': 613, 'Philadelphia': 482, 'Washington': 396},
    'Boston': {'Detroit': 613, 'New York': 190},
    'New York': {'Boston': 190, 'Philadelphia': 81},
    'Philadelphia': {'New York': 81, 'Detroit': 482, 'Washington': 123},
    'Washington': {'Detroit': 396, 'Philadelphia': 123, 'Atlanta': 543, 'Miami': 923},
    'Atlanta': {'Chicago': 588, 'Dallas': 721, 'Houston': 702, 'Washington': 543, 'Miami': 604},
    'Miami': {'Houston': 968, 'Atlanta': 604, 'Washington': 923}
}

print(f"Graph loaded successfully with {len(graph)} cities.")

# Heuristic estimates h(n) towards Goal: Miami
# Admissible condition: h(n) <= true remaining distance

heuristics_miami = {
    'Seattle': 2700,
    'San Francisco': 2500,
    'Los Angeles': 2300,
    'Riverside': 2250,
    'Phoenix': 1900,
    'Dallas': 1100,
    'Houston': 900,
    'Chicago': 1200,
    'Detroit': 1100,
    'Boston': 1200,
    'New York': 1100,
    'Philadelphia': 1000,
    'Washington': 900,
    'Atlanta': 500,
    'Miami': 0
}

import heapq

def a_star_search(graph, heuristics, start, goal):
    """
    Core A* Search Algorithm (Member 2 Module)
    - pq: Priority Queue holding tuples of (f_score, current_node)
    - g_score: Cost from start node to current node
    - came_from: Parent pointer mapping for path reconstruction
    """
    # 1. Initialize Priority Queue with start node
    pq = []
    heapq.heappush(pq, (heuristics[start], start))

    # 2. Initialize g_scores with infinity
    g_score = {node: float('inf') for node in graph}
    g_score[start] = 0

    # 3. Parent tracking dictionary
    came_from = {}

    # 4. Closed set to avoid re-expanding visited nodes
    closed_set = set()

    while pq:
        # Pop node with lowest f(n)
        current_f, current = heapq.heappop(pq)

        # Goal state check
        if current == goal:
            # Reconstruct shortest path
            path = []
            curr = goal
            while curr in came_from:
                path.append(curr)
                curr = came_from[curr]
            path.append(start)
            path.reverse()
            return path, g_score[goal]

        if current in closed_set:
            continue
        closed_set.add(current)

        # Explore neighbors
        for neighbor, weight in graph[current].items():
            tentative_g = g_score[current] + weight

            # If a shorter path to neighbor is found
            if tentative_g < g_score[neighbor]:
                g_score[neighbor] = tentative_g
                came_from[neighbor] = current
                f_score = tentative_g + heuristics.get(neighbor, 0)
                heapq.heappush(pq, (f_score, neighbor))

    return None, float('inf')

    # Execute A* Search from Seattle to Miami
start_city = 'Seattle'
goal_city = 'Miami'

path, total_cost = a_star_search(graph, heuristics_miami, start_city, goal_city)

# Required Outputs
print("=" * 50)
print(f"1) Shortest Path: {' -> '.join(path)}")
print(f"2) Cost of Shortest Path: {total_cost} miles")
print(f"3) Cities Along the Path: {', '.join(path[1:-1])}")
print("=" * 50)

# Heuristic estimates h(n) towards Goal: New York
# These are approximate straight-line distances or informed guesses.
heuristics_new_york = {
    'Seattle': 2400,
    'San Francisco': 2900,
    'Los Angeles': 2800,
    'Riverside': 2750,
    'Phoenix': 2100,
    'Dallas': 1600,
    'Houston': 1600,
    'Chicago': 750,
    'Detroit': 500,
    'Boston': 200,
    'New York': 0,
    'Philadelphia': 80,
    'Washington': 200,
    'Atlanta': 750,
    'Miami': 1100
}

start_city_benchmark = 'San Francisco'
goal_city_benchmark = 'New York'

path_benchmark, total_cost_benchmark = a_star_search(graph, heuristics_new_york, start_city_benchmark, goal_city_benchmark)

print("=" * 50)
print(f"1) Shortest Path (SF to NY): {' -> '.join(path_benchmark) if path_benchmark else 'No path found'}")
print(f"2) Cost of Shortest Path (SF to NY): {total_cost_benchmark} miles")
if path_benchmark:
    print(f"3) Cities Along the Path (SF to NY): {', '.join(path_benchmark[1:-1])}")
else:
    print("3) Cities Along the Path (SF to NY): N/A")
print("=" * 50)
```
